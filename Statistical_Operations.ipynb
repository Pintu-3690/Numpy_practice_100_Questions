```python
#Find the sum of all elements.
import numpy as np
a = np.array([2,5,8,7,9,6])
sum = np.sum(a)
print(sum)
```

```python
#Find the mean of an array.
m = np.mean(a)
print(m)
```

```python
#Find the median of an array.
me = np.median(a)
print(me)
```

```python
#Find the minimum value.
min = np.min(a)
print(min)
```

```python
#Find the maximum value.
max = np.max(a)
print(max)
```

```python
#Find the standard deviation.
st = np.std(a)
print(st)
```

```python
#Find the variance.
var = np.var(a)
print(var)
```

```python
#Find the index of the minimum element.
min_index = np.argmin(a)
print(min_index)
```

```python
#Find the index of the maximum element.
max_index = np.argmax(a)
print(max_index)
```

```python
#Calculate the cumulative sum.
cum_sum = np.cumsum(a)
print(cum_sum)
```
