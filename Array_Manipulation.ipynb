```python
#Reshape a 1D array into a 2D array.
import numpy as np
a = np.array([1,2,3,4,5,6])
print(a)
print(a.shape)
print(a.reshape(2,3))
print(a.reshape(2,3).shape)
```

```python
#Convert a 2D array into a 1D array.
print(a.reshape(1,6))
```

```python
#Transpose a matrix.
b = np.array([[1,2,3],[4,5,6]])
print(b)
print(b.T)
print(b.T.shape)
```

```python
#Flatten a matrix.
print(b.flatten())
print(b.flatten().shape)
```

```python
#Swap the rows and columns of a matrix.
c =np.array([[1,2,3],[4,5,6]])
print(c)
print(c.T)
```
